```python
import numpy as np
```

```python
# Create two NumPy arrays
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 5, 8])
```

```python
# Summation
summation = np.sum([arr1, arr2], axis=0)

# Product
product = np.prod([arr1, arr2], axis=0)

# Difference
difference = np.diff([arr1, arr2], axis=0)
```

```python
print("Summation:", summation)
print("Product:", product)
print("Difference:", difference)
```
